```python
from choice_learn.datasets import load_expedia
from choice_learn.models import ConditionalLogit
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
df = load_expedia(as_frame=False, preprocessing="rumnet")
```

## Train test split

```python
N = len(df)
rng = np.random.default_rng(42)
idx = rng.permutation(N)

n_train = int(0.8*N)
train_idx = idx[:n_train]
test_idx = idx[n_train: ]

#train and test data set
train_ds = df[train_idx]
test_df = df[test_idx]
```

```python
train_ds.summary()
```

The dataset contains 318,094 choice situations and 39 items (hotels) per choice set. It includes 13 shared (context) features describing the search characteristics and 11 item-specific features describing the hotel attributes.

```python
#specifiying the model 
model = ConditionalLogit(
    optimizer="Adam",   
   epochs=50,
    batch_size=512,   
    lr=0.005           
)


J = 39
base_alt = 0 

#fixed effects
model.add_coefficients("intercept", items_indexes=[j for j in range(J) if j != base_alt])


#adding shared featuers
model.add_shared_coefficient(feature_name="log_price",items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="orig_destination_distance", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="prop_starrating", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="prop_review_score", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="prop_brand_bool", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="prop_location_score1", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="prop_location_score2", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="srch_length_of_stay", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="srch_adults_count", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="srch_children_count", items_indexes=list(range(J)))
model.add_shared_coefficient(feature_name="srch_room_count", items_indexes=list(range(J)))

```

```python
history = model.fit(train_ds, verbose=1,get_report=True)

plt.figure()
plt.plot(history["train_loss"])
plt.xlabel("Epoch")
plt.ylabel("Cross-entropy (NLL)")
plt.title("Training loss")
plt.show()
```

```python
#evaluating the model
p_test = model.predict_probas(test_df)
y_test = tf.convert_to_tensor(test_df.choices, dtype=tf.float32)
cce = tf.keras.losses.CategoricalCrossentropy(from_logits=False)
test_ce = cce(y_test, p_test).numpy()
print("Test cross-entropy:", test_ce)
print("Choice-Learn evaluate() (mean NLL):", model.evaluate(test_ds).numpy())
```
